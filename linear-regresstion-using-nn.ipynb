```python
t_c = [0.5,  14.0, 15.0, 28.0, 11.0,  8.0,  3.0, -4.0,  6.0, 13.0, 21.0]
t_u = [35.7, 55.9, 58.2, 81.9, 56.3, 48.9, 33.9, 21.8, 48.4, 60.4, 68.4]
t_c = torch.tensor(t_c).unsqueeze(1)
t_u = torch.tensor(t_u).unsqueeze(1)
```

```python
len_sample = t_c.shape[0]
val_len = int(len_sample * 0.2)

suffled_index = torch.randperm(len_sample)

x_index = suffled_index[val_len:]
val_index = suffled_index[:val_len]

X, Xt = t_c[x_index], t_u[x_index]
V, Vt = t_c[val_index], t_u[val_index]
```

```python
sequenctial_model = nn.Sequential( 
    nn.Linear(1,15),
    nn.ReLU(),
    nn.Linear(15,1),
    
)
optimizer = optim.SGD(
    sequenctial_model.parameters(),
    lr=1e-4,
)

def loss_fn(p, t):
    return (p-t) ** 2
```

```python
def training(epoch, optimizer, X, Xt, V, Vt, model):
    for i in range(epoch):
        X_p = model(X)
        loss_X = loss_fn(X_p, Xt).mean()
        V_p = model(V)
        loss_V = loss_fn(V_p, Vt).mean()

        optimizer.zero_grad()
        loss_X.backward()
        optimizer.step()

        if (epoch % 10) == 0:
            print(f"Training Loss:{loss_X}, Validation Loss:{loss_V}") 


training(5000, optimizer, X, Xt, V, Vt, sequenctial_model)
```
